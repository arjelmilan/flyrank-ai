## Signal test #1 : content_age_days

*Three safe signals, each with a mini-test and a verdict: CONFIRMED / OPPOSITE / MIXED / FALSE.*

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/arjelmilan/flyrank-ai/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

df['click_drop'] = np.maximum(0,df["clicks_prev_30d"] - df["clicks_last_30d"])
df['decay_event'] = (df['click_drop']>0).astype(int)

age_bucket = df.groupby('freshness_tier',observed=False).agg(
    sample_size_n = ("content_id","count"),
    mean_click_drop=("click_drop", "mean"),
    decay_rate=("decay_event", "mean")
).reset_index()
print("=== SIGNAL 1: FRESHNESS TIER AUDIT ===")
print(age_bucket.to_string(index=False))
print("Verdict: CONFIRMED — Older content (>.91 days) shows higher decay rates.")
```

## signal_test2: position vs ctr

```python

pos_bucket = df.groupby('position_tier',observed=False).agg(
    sample_size_n = ("content_id","count"),
    mean_ctr=("ctr", "mean"),
    mean_click_drop =("click_drop", "mean")
).reset_index()
print("\n=== SIGNAL 2: POSITION TIER AUDIT ===")
print(pos_bucket.to_string(index=False))
print("Verdict: CONFIRMED — Positions 4-10 with low CTR represent prime refresh targets.")
```
